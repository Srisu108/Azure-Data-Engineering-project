```python
#create spearte dataframes for the silver layer.
# The originial dataframes remain unchanged.

silver_dataframes ={}

silver_dataframes["cast"] = dataframes["cast"].drop_duplicates().copy()
silver_dataframes["category"] = dataframes["category"].drop_duplicates().copy()
silver_dataframes["countries"] = dataframes["countries"].drop_duplicates().copy()
silver_dataframes["directors"] = dataframes["directors"].drop_duplicates().copy()

# validate the results
for name in silver_dataframes:
    print(
        name,
        "|bronze/local source rows:", len(dataframes[name]),
        "|silver rows:", len(silver_dataframes[name]),
        "|Duplicates remaining:", silver_dataframes[name].duplicated().sum()
    )
```

```python
from pathlib import Path
output_dir = Path("output")
output_dir.mkdir(parents=True, exist_ok=True)

for name, df in silver_dataframes.items():
    output_file = output_dir / f"netflix_{name}.csv"
    df.to_csv(output_file, index=False)
    print(f"save: {output_file} | Rows:{len(df)}")
```

```python
##### Acting don the Netflix_title.csv files for transformations.

titles_df = pd.read_csv(data_dir / "netflix_titles.csv")
print("Shape:", titles_df.shape)
print("\ncolumns:")
print(titles_df.dtypes)

print("\nMissing values:")
print(titles_df.isnull().sum())

print("\nFirst 5 rows:")
print(titles_df.head())
```

```python
print("Content types:")
print(titles_df["type"].value_counts(dropna=False))

print("\nMissing duration values by content type:")
print(
    titles_df.groupby("type", dropna=False)[
        ["duration_minutes", "duration_seasons"]
    ].agg(lambda column:column.isna().sum())
)

print("\nRows missing key identifiers or titles:")
display(
    titles_df[
        titles_df["show_id"].isna()
        | titles_df["title"].isna()
        | titles_df["type"].isna()
    ]
)
```

```python
####Step 2 — Create the working DataFrame and flag exceptions


silver_titles = titles_df.copy()

silver_titles["quality_issue"] = (
    ~silver_titles["type"].isin(["Movie", "TV Show"])
    | silver_titles["title"].isna()
    | silver_titles["show_id"].isna()
)

print("Total source rows:", len(titles_df))
print("Rows flagged for review:", silver_titles["quality_issue"].sum())
print("\nContent types, including invalid values:")
print(silver_titles["type"].value_counts(dropna=False))
```

```python
silver_titles["type_flag"] = 0
silver_titles.loc[silver_titles["type"]== "Movie", "type_flag"] = 1
silver_titles.loc[silver_titles["type"]=="TV Show", "type_flag"] = 2 

print(silver_titles[["type", "type_flag"]].value_counts(dropna=False))
```

```python
silver_titles["duration_minutes"] = pd.to_numeric(
    silver_titles["duration_minutes"], errors="coerce"
)

silver_titles["duration_seasons"] = pd.to_numeric(
    silver_titles["duration_seasons"], errors="coerce"
)

print(silver_titles[["duration_minutes", "duration_seasons"]])

```

```python
print(silver_titles[["duration_minutes", "duration_seasons"]].dtypes)

print("\nMissing values after conversion")
print(silver_titles[["duration_minutes", "duration_seasons"]].isna().sum())
```

```python
for column in ["duration_minutes", "duration_seasons"]:
    original_values = titles_df[column]
    converted_values = silver_titles[column]

    failed = original_values.notna() & converted_values.isna()

    print(f"\n{column}:")
    print("Failed conversions:", failed.sum())
    print("Original values:")
    print(original_values[failed].tolist())
```

```python
silver_titles["title_prefix"] = (
    silver_titles["title"]
    .str.split(":", n=1)
    .str[0]
    .str.strip()
)
display(silver_titles[["title", "title_prefix"]].head(10))
```

```python
print(silver_titles["rating"].value_counts(dropna=False))
```

```python
silver_titles["rating_prefix"] = (
    silver_titles["rating"]
    .str.split("-", n=1).str[0]
)
display(silver_titles[["rating", "rating_prefix"]].head(10))
```

```python
# Restore duration values from the original dataset

silver_titles["duration_minutes"] = pd.to_numeric(
    titles_df["duration_minutes"], errors = "coerce"
)

silver_titles["duration_minutes"] = pd.to_numeric(
    titles_df["duration_minutes"], errors = "coerce"
)

# Validate the restored values

display(
    silver_titles[["type", "title", "duration_minutes", "duration_seasons" ]].head(5)
)
```

```python
movie_mask = silver_titles["type"] == "Movie"

silver_titles["duration_rank"] = pd.NA

silver_titles.loc[movie_mask, "duration_rank"] = (
    silver_titles.loc[movie_mask, "duration_minutes"]
    .rank(method="dense", ascending=False)
)

display(
    silver_titles.loc[
        movie_mask & silver_titles["duration_minutes"].notna(),
        ["title", "duration_minutes", "duration_rank"]
    ]
    .sort_values("duration_rank")
    .head(10)
)
```

```python
type_summary = (
    silver_titles
    .groupby("type", dropna=False)
    .size()
    .reset_index(name="record_count")
)

display(type_summary)
```

```python
silver_titles["title_prefix"] = (
    titles_df["title"]
    .str.split(":", n=1)
    .str[0]
    .str.strip()
)

display(
    silver_titles[["title", "title_prefix"]].head(10))
```

```python
print("Original row count", len(titles_df))
print("silver row count", len(silver_titles))

print("\n Dervied columns:")
print(silver_titles[
    ["type", "type_flag", "title_prefix", "rating_prefix", "duration_rank"]
].head(10))

print("\nQuality issues flagged:", silver_titles["quality_issue"].sum())
```

```python

print("=== SILVER VALIDATION ===")

print("Source rows:", len(titles_df))
print("Silver rows:", len(silver_titles))
print("Row count preserved:", len(titles_df) == len(silver_titles))

print("\nMissing title prefixes:", silver_titles["title_prefix"].isna().sum())
print("Missing type flags:", silver_titles["type_flag"].isna().sum())
print("Quality issues flagged:", silver_titles["quality_issue"].sum())

print("\nDuration rank summary:")
print(silver_titles.loc[
    silver_titles["type"] == "Movie", "duration_rank"
].describe())
```

```python

silver_titles["duration_rank"] = pd.to_numeric(
    silver_titles["duration_rank"], errors="coerce"
)

print("Rank data type:", silver_titles["duration_rank"].dtype)

print("\nRanking summary:")
print(
    silver_titles.loc[
        silver_titles["type"] == "Movie", "duration_rank"
    ].describe()
)
```
